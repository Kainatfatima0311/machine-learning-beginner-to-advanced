```python
# Import required library
import pandas as pd
```

```python
#load dataset
df = pd.read_csv(
    "../heart_failure_clinical_records_dataset-selected-columns.csv"
)
```

```python
#display basic data
print("FIRST 5 ROWS")
print(df.head())


print("LAST 5 ROWS")
print(df.tail())
```

```python
#checking data dimensions
print("Complete Shape:", df.shape)
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
```

```python
#display columns 
for index, column in enumerate(df.columns, start=1):
    print(f"{index}. {column}")
```

```python
#datatypes
print(df.dtypes)
```

```python
#display dataset info
df.info()
```

```python
#identify numerical columns
numerical_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()
for column in numerical_columns:
    print("-", column)
```

```python
#identify binary columns 
binary_categorical_columns = [
    "anaemia",
    "diabetes",
    "high_blood_pressure",
    "sex"
]


for column in binary_categorical_columns:
    print(df[column].value_counts().sort_index())
```

```python
#identify continuous measurement column
continuous_columns = [
    "age",
    "creatinine_phosphokinase",
    "ejection_fraction",
    "platelets",
    "serum_creatinine",
    "serum_sodium"
]
for column in continuous_columns:
    print("-", column)
```

```python
#statiscal summary
print(df.describe())
```

```python
#checking unique values
print(df.nunique())
```

```python
#chceking missing values
missing_values = df.isnull().sum()
print(missing_values)
#
```

```python
#chcking duplicate values
df.duplicated().sum()
```

```python
#display summary of dataset
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
print(f"Missing Values: {df.isnull().sum().sum()}")
print(f"Duplicate Rows: {df.duplicated().sum()}")
print(f"Numerical Columns: {len(numerical_columns)}")
print(f"Binary Categorical Columns: {len(binary_categorical_columns)}")
print(f"Continuous Columns: {len(continuous_columns)}")
```
