```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the Heart Failure dataset
df = pd.read_csv(
    "../heart_failure_clinical_records_dataset-selected-columns.csv"
)

# Display first 5 rows
df.head()
```

```python
# Select measurement-based numerical features

numerical_features = [
    "age",
    "creatinine_phosphokinase",
    "ejection_fraction",
    "platelets",
    "serum_creatinine",
    "serum_sodium"
]

numerical_features
```

```python
# Create histograms for numerical features

df[numerical_features].hist(
    figsize=(15, 10),
    bins=15,
    edgecolor="black"
)

plt.suptitle(
    "Distribution of Numerical Features",
    fontsize=16,
    y=1.02
)

plt.tight_layout()
plt.show()
```

```python
# Scatter Plot: Age vs Serum Creatinine

plt.figure(figsize=(8, 6))

plt.scatter(
    df["age"],
    df["serum_creatinine"],
    alpha=0.7
)

plt.title("Age vs Serum Creatinine")
plt.xlabel("Age")
plt.ylabel("Serum Creatinine")
plt.grid(alpha=0.3)

plt.show()
```

```python
# Pair Plot for selected numerical features

selected_features = [
    "age",
    "ejection_fraction",
    "serum_creatinine"
]

sns.pairplot(
    df[selected_features],
    diag_kind="hist"
)

plt.show()
```
