# Model Selection Basics

## Objective

In this notebook, we will practice identifying:

- Binary Classification
- Multiclass Classification
- Regression

The goal is to understand the target variable before selecting a Machine Learning model.

```python
import pandas as pd
```

# Binary Classification

Binary Classification contains exactly two possible target classes.

```python
binary_data = pd.DataFrame({
    "age": [22, 35, 47, 29, 51],
    "income": [30000, 55000, 80000, 42000, 90000],
    "loan_approved": [0, 1, 1, 0, 1]
})

binary_data
```

```python
binary_target = binary_data["loan_approved"]

print(
    "Target Classes:",
    binary_target.unique()
)

print(
    "Number of Classes:",
    binary_target.nunique()
)
```

# Multiclass Classification

Multiclass Classification contains more than two possible target classes.

```python
multiclass_data = pd.DataFrame({
    "weight": [150, 120, 180, 130, 160, 110],
    "color_score": [8, 7, 5, 9, 6, 8],
    "fruit": [
        "Apple",
        "Banana",
        "Orange",
        "Apple",
        "Orange",
        "Banana"
    ]
})

multiclass_data
```

```python
multiclass_target = multiclass_data["fruit"]

print(
    "Target Classes:",
    multiclass_target.unique()
)

print(
    "Number of Classes:",
    multiclass_target.nunique()
)
```

# Regression

Regression predicts a continuous numerical value.


```python
regression_data = pd.DataFrame({
    "area": [1000, 1500, 1800, 2200, 2500],
    "bedrooms": [2, 3, 3, 4, 4],
    "house_price": [
        5000000,
        7500000,
        8500000,
        10500000,
        12000000
    ]
})

regression_data
```

```python
comparison = pd.DataFrame({
    "Problem Type": [
        "Binary Classification",
        "Multiclass Classification",
        "Regression"
    ],
    "Target Example": [
        "0 / 1",
        "Apple / Banana / Orange",
        "House Price"
    ],
    "Question": [
        "Which of 2 classes?",
        "Which of multiple classes?",
        "How much?"
    ]
})

comparison
```

# Summary

## Binary Classification
- Predicts exactly two classes.
- Example: Loan Approved / Not Approved.

## Multiclass Classification
- Predicts more than two classes.
- Example: Apple / Banana / Orange.

## Regression
- Predicts a continuous numerical value.
- Example: House Price.

## Quick Rule

- Two classes → Binary Classification
- More than two classes → Multiclass Classification
- Continuous numerical target → Regression

Always understand the meaning of the target before selecting a Machine Learning algorithm.
